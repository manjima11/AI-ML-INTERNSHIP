Mini Project: Student Result System with Search 

Develop a console-based Python program that manages student results. The program should allow users to input student names and marks, calculate averages, assign grades based on conditions, display all student results, and provide a search feature to find a specific student by name.

```python
# Student Result System

# Function to calculate grade
def calculate_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 75:
        return "A"
    elif avg >= 60:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "F"

# Input number of students
n = int(input("Enter number of students: "))

students = []

# Input data
for i in range(n):
    print(f"\nEnter details for student {i+1}")
    name = input("Name: ")

    marks = []
    for j in range(3):
        m = float(input(f"Enter marks for subject {j+1}: "))
        marks.append(m)

    total = sum(marks)
    avg = total / 3
    grade = calculate_grade(avg)

    student = {
        "name": name,
        "marks": marks,
        "total": total,
        "average": avg,
        "grade": grade
    }

    students.append(student)

# Display all student results
print("\n--- All Student Results ---")
for s in students:
    print("\nName:", s["name"])
    print("Marks:", s["marks"])
    print("Total:", s["total"])
    print("Average:", round(s["average"], 2))
    print("Grade:", s["grade"])

# Search feature
search_name = input("\nEnter student name to search: ")

found = False
for s in students:
    if s["name"].lower() == search_name.lower():
        print("\n--- Student Found ---")
        print("Name:", s["name"])
        print("Marks:", s["marks"])
        print("Total:", s["total"])
        print("Average:", round(s["average"], 2))
        print("Grade:", s["grade"])
        found = True
        break

if not found:
    print("Student not found.")
```
